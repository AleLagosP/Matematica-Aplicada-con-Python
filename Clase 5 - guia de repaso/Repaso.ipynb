## 1 Respuesta

```python
def S(x):
    return -1.6*x**2 + 1.3*x + 141.3

def R(x):
    return 1.4*x**2 - 5.6*x + 12.6

x = 1

print(f'La función en el punto x es {S(x):.2f}.')
print(f'La función en el punto x es {R(x):.2f}.')
```

```python
# Velocidad del automóvil
def s(x):
    return -1.6*x**2 + 1.3*x + 141.3

# Velocidad de la moto
def r(x):
    return 1.4*x**2 - 5.6*x + 12.6

# Diferencia entre ambas velocidades
def diferencia(x):
    return s(x) - r(x)

# Buscar los puntos donde s(x) = r(x)
x1 = fsolve(diferencia, -5)[0]
x2 = fsolve(diferencia, 8)[0]

print(f"Intersecciones: {x1:.1f} y {x2:.1f} minutos")

# Como el tiempo no puede ser negativo:
print(f"El automóvil es más rápido entre 0 y {x2:.1f} minutos")
```
