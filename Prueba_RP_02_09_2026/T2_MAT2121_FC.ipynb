## Respuesta 2

```python
#Grafico de la funcion lineal

def A(x):
  return 0.66*x + -11.44

def G(x):
  return 0.01*x**2 - 0.4*x + 8 

x = np.arange(20, 124, 2)  #  valores que python crea la función y de cuanto en cuanto en cuanto va

plt.plot(x, A(x), color = 'r', label = 'Metodo A H(x)')
plt.plot(x, G(x), color = 'g', label = 'Metodo B G(x)')

plt.ylabel('y = Tamaño de Encriptacion')   # nombre eje y
plt.xlabel('x = tiempo(segundos)')   # nombre eje x

plt.ylim(-20, 90)   # limites del eje y que quieres ver
plt.xlim(20, 90)   # limites  del eje x que quieres ver

plt.title('Cantidad de archivos encriptados en función del tiempo')  # titulo del grafico

plt.grid()
plt.legend()
plt.show()
```

```python
def f(x):
  return G(x)-A(x)

x = np.linspace(0, 100, 10)  # crea puntos de partida  para  para que solve busque las respuestas

#np.around() permite redondear los valores a 2 decimales y elimina los restantes
solucion = np.around(fsolve(f, x), decimals = 1)

#np.unique() permite eliminar las soluciones repetidas
interseccion = np.unique(solucion)

print(f'Los puntos de intersección de las funciones están en x = {interseccion}')  
```
