```python
#Determinando las ciudades del primera hora de lluvia a las 6 de la mañana:

#1. valparaiso
def V(t):
    return 0.9*t**2 + 12.5*t +36

print(f'El valor de Tiempo es {V(0):.0f} Milimetros cada hora')

# 2. Santiago
def S(t):
    return 0.95*t**2 + 16.8*t +31
print(f'El valor de Tiempo es {S(0):.0f} Milimetros cada hora')

#3. Comcepcion 
def C(t):   
    return 1.1*t**2+ 19.4*t + 26
print(f'El valor de Tiempo es {C(0):.0f} Milimetros cada hora')
```

```python
#Pregunta 2, Grafico

def V(t):
    return 0.9*t**2 + 12.5*t +36

def S(t):
    return 0.95*t**2 + 16.8*t +31

def C(t):
    return 1.1*t**2+ 19.4*t + 26

t= np.arange(0, 15 ,1)

plt.plot(t, V(t), label='Valparaiso')
plt.plot(t, S(t), label='Santiago')
plt.plot(t, C(t), label='Concepcion')
plt.plot(14, V(14), 'ro', label=f'Resultado V= {V(10):.0f} (mm)')
plt.plot(14, S(14), 'bo', label=f'Resultado S = {S(10):.0f} (mm)')
plt.plot(14, C(14), 'go', label=f'Resultado C = {C(10):.0f} (mm)')
plt.title('Pronostico del Clima de cada ciudad')
plt.ylabel('Y = Cantidad de lluvia(Milimetros)')
plt.xlabel('X = Tiempo (Horas)')
plt.legend()
plt.grid()
plt.show()
```
