```python
import numpy as np                  # Cálculos numéricos, vectores y matrices
import pandas as pd                 # Manipulación y análisis de bases de datos
import matplotlib.pyplot as plt     # Creación de gráficos
from scipy.optimize import fsolve   # Resolución numérica de ecuaciones
```

# Enunciado

En un experimento de biología se estudia el crecimiento de dos tipos de plantas, durante un período de 9 a 80 días.

La Planta A presenta un crecimiento aproximadamente constante. Por cada 2 días, su altura aumenta 0,8 cm, como se muestra en la tabla.

La altura de la Planta B, en cambio, varía de acuerdo con un modelo cuadrático.

#### Método A:
  $$f(x) = 0,4*x +3,4$$

#### Método B:


$$g(x)=0,01x^2-0,3x+9,7$$

Con ayuda de Python resuelve las siguientes preguntas, considera un decimal para responder:

A_¿Qué tipo de función permite representar el crecimiento de la Planta A? interprete la PENDIENTE


B_Calcule la altura de ambas plantas cuando han transcurrido 27 días. ¿Cuál planta presenta mayor altura?


C_¿En qué día(s) ambas plantas alcanzan la misma altura? ¿Cuál es esa altura? 


D¿Durante qué período la Planta A presenta una altura mayor que la Planta B? (Recuerda GRAFICAR)





# Desarrollo

```python
#1 metodo A: F(x) = 0,4*x + 3,4
def F(x):
    return 0.4 * x + 3.4

x = 27 # dias

print(f'La Planta A presenta un crecimiento de 17 dias con una altura de {F(x):.1f} cm')

#-------------------------------------------------------------------------------------------------------------------------------------

# 2 Metodo B: G(x) = 0,01*x**2 - 0,3*x + 9.7
def G(x):
    return 0.01*x**2 - 0.3*x + 9.7

x = 27 # dias

print(f'La Planta B presenta un crecimiento de 17 dias con una altura de {G(x):.1f} cm ')
```

```python
def f(x):
    return 0.4 * x + 3.4

#Valor/es inicial/es de la aproximación
xo = np.linspace(0, 100, 2)

solucion = fsolve(f, xo)

print(f'La funcion pendiente representa {solucion[0]:.0f} dias de crecimiento de la Planta A')
```

```python
def G(x):
    return 0.01*x**2 - 0.3*x + 9.7

#Valor/es inicial/es de la aproximación
xo = np.linspace(0, 100, 2)

solucion = fsolve(G, xo)

print(f'La funcion pendinete representa {solucion[0]:.0f} dias de crecimiento de la Planta B')
```

```python
# las funciones F y G

def A(x):
  return F(x) - G(x)

x = np.linspace(0.4, 100, 10)  # crea puntos de partida  para  para que solve busque las respuestas

#np.around() permite redondear los valores a 2 decimales y elimina los restantes
solucion = np.around(fsolve(A, x), decimals = 2)

#np.unique() permite eliminar las soluciones repetidas
interseccion = np.unique(solucion)

print(f'Los puntos de intersección de las funciones están en x = {interseccion}')    # cuando son varias
```

```python
#Grafico de las funciones F y G
def h(x):
  return F(x)

def g(x):
  return G(x)

x = np.arange(0.8, 100, 0.001)  #  valores que python crea la función y de cuanto en cuanto en cuanto va

plt.plot(x, h(x), color = 'r', label = 'h(x)')
plt.plot(x, g(x), color = 'g', label = 'g(x)')

plt.ylabel(' y')   # nombre eje y
plt.xlabel(' x')   # nombre eje x

plt.ylim(0, 100)   # limites del eje y que quieres ver
plt.xlim(0, 100)   # limites  del eje x que quieres ver

plt.title('Comparación entre las funciones h(x) y g(x)')

plt.grid()
plt.legend()
plt.show()
```
