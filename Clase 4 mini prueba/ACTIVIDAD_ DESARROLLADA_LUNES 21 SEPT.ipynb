```python
2
#Crecimiento planta A para 27 días
def a(x):
    return 0.4*x+3.4
x1= 27
print(f'Para 27 días la planta A tiene una altura de {a(x1):.1f} cm')

#Crecimiento palnta B para 27 días
def b(x):
    return 0.01*x**2-0.3*x+9.7
x2= 27
print(f'Para 27 días la planta B tiene una altura de {b(x2):.1f} cm')
    
```

```python
3
def a(x):
  return 0.4*x+3.4

def b(x):
  return 0.01*x**2-0.3*x+9.7

x = np.arange(0, 100, 0.01)  #  valores que python crea la función y de cuanto en cuanto en cuanto va
                                           
plt.plot(x, a(x), color = 'r', label = 'a(x)')
plt.plot(x, b(x), color = 'g', label = 'b(x)')

plt.ylabel(' Altura (cm)')   # nombre eje y
plt.xlabel(' Tiempo (días)')   # nombre eje x

plt.ylim(-1, 50)   # limites del eje y que quieres ver
plt.xlim(0, 90)   # limites  del eje x que quieres ver

plt.title('Comparación entre los crecimientos de la plana A y B con respecto a los días')

plt.grid()
plt.legend()
plt.show()
```

```python
3
#Definiendo la función igualada a cero
def f(x):
  return b(x)-a(x)

x = np.linspace(0, 100, 10)  # crea puntos de partida  para  para que solve busque las respuestas

#np.around() permite redondear los valores a 2 decimales y elimina los restantes
solucion = np.around(fsolve(f, x), decimals = 1)

#np.unique() permite eliminar las soluciones repetidas
interseccion = np.unique(solucion)

print(f'Los puntos de intersección de las funciones están en x = {interseccion}')    # cuando son varias
```
