## Pregunta A
Un nuevo software tiene $500$ usuarios y se espera que el número de usuarios crezca un $15\%$ respecto al mes anterior. Considere que $b_{n}$ es el número de usuarios transcurridos $n$ meses.

1. Calcula cuántos usuarios tendrá el software al cabo de tres meses. ¿Y al cabo de seis meses?
2. Mediante un código en *Pyhton* muestra la cantidad de usuarios que tendrá el software cada mes durante el primer año.
2. ¿Cuántos usuarios tendrá en total el software luego de un año? Utilice ciclos *for* para calcular.

## 1. SUCESIÓN GEOMMÉTRICA:

```python
# SUCESIÓN GEOMMÉTRICA
#Término general sucesión GEOMÉTRICA 
#RAZÓN CONSTANTE  (DUPLICAR, TRIPLICAR , %)  r

a_3 = 500 * 1.15**(3)

a_6 = 500 * 1.15**(6)

print(f'Al cabo de .... habrá {a_3:.0f} ....')

print(f'Al cabo de .... habrá {a_6:.0f} ....')
```

## 2. Crear una secesion 

```python
# otra forma de crear una secesion   
# EXPLICACION

sucesion = []  # Crea una lista vacía para guardar los términos de la sucesión

for n in range(1, 13):  # Recorre los valores de n desde inicio hasta fin
    sucesion.append(round(500 * 1.15**n))  # Calcula cada término y lo agrega a la lista

print(sucesion)  # Muestra en pantalla la lista con los términos calculados
```

## 3. Suma 

```python
# Suma de términos de una lista      

suma = 0    # Inicializa la suma en cero

for n in range(1, 12):  # Recorre los primeros n términos
    suma = suma +  500 * 1.15**n  # Va sumando cada número 

print(f'La suma ... {suma:.0f}')  # Muestra en pantalla la suma total de los términos
```


## Pregunta 2

Martina decidió que dejará de aumentar sus depósitos cuando el monto correspondiente a una semana alcance los **$100.000**.

**¿En qué semana su depósito alcanzará por primera vez los $100.000?**

Utiliza Python para determinar la posición correspondiente dentro de la sucesión e interrpete el resultado

## 1. Determinar si un determinado valor es término de la sucesión.  y la POSICIÓN DEL TERMIN0

```python
#Determinar si un determinado valor es término de la sucesión. Y POSICION DEL TERMINO
final= 100000
termino=0
n=0
dif=final-termino
while dif > 0:
    n += 1
    termino= 5000*n + 10000
    dif=final-termino

if dif==0:
    print(f'El {final} es un término de la sucesión y ocupa el lugar {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```
